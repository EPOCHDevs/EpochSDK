## 3. Read and check the saved table

These reads start no model or engine job. The arithmetic checks use actual returned values; they are separate from source/vintage and trace review.

```python
import math
tables = await read(epoch, result)
rows = tables["gdp_revisions"]
assert rows and len({r["quarter"] for r in rows}) == len(rows)
for row in rows:
    assert row["quarter"].startswith(("2023", "2024"))
    assert math.isclose(row["revision_pp"], row["latest_growth"] - row["first_growth"], rel_tol=1e-8, abs_tol=1e-8)
print("Saved quarter labels and revision arithmetic checked; review the actual source definitions and clocks.")
```
