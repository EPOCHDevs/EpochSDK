## 3. Read and check the saved table

These reads start no model or engine job. The arithmetic checks use actual returned values; they are separate from source/vintage and trace review.

```python
import math
tables = await read(epoch, result)
rows = tables["daily_returns"]
assert rows and [r["date"] for r in rows] == sorted(r["date"] for r in rows)
assert all("2026-09-01" <= r["date"] <= "2026-09-30" and r["close"] > 0 for r in rows)
for previous, current in zip(rows, rows[1:]):
    assert current["daily_return"] is not None
    assert math.isclose(current["daily_return"], current["close"] / previous["close"] - 1, rel_tol=1e-8, abs_tol=1e-10)
print("Saved dates, ordering and return arithmetic checked.")
```
