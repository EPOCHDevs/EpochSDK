# QQQ, SPY and Treasury yields

How did QQQ perform versus SPY in months when US 10-year yields rose sharply, fell sharply or stayed fairly flat during 2016–2025?

**Data and assumptions**

- US 10-year nominal constant-maturity Treasury yields (Fed H.15 definition) and dividend-inclusive QQQ/SPY returns.
- Calendar months January 2016–December 2025, with December 2015 as the baseline; align yields to the last equity session of each month.
- Rising means a same-month yield increase of at least 25 bp; falling means a decrease of at least 25 bp; otherwise flat. This is a descriptive comparison, not a trading signal.

**Requested answer:** rate_regimes (table), macro_assessment (markdown).
