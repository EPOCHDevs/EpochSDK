Use an existing completed result. This question does not ask for a saved or reusable report: Epoch chooses the supported native path. No financial values are prefilled. Share this folder with its helpers and cases. See [README](README.md) for login, service URLs and source IDs.

## 1. Connect

This prepares the ordinary SDK client. It does not start work. Use `epoch-sdk[notebook]` 0.1.0a25 or later and native SDK 3.3.0rc19 or later.

```python
from report_helpers import connect, task, display_request, run, read, native_operation, dashboard
from epoch_sdk.research import ResearchCompleted

epoch = connect()
```

## 2. Review the question and outputs

The source ID comes from your prepared environment. Keep the attempt to reconnect; change it for intentionally changed inputs. The case JSON beside the notebook contains the exact task, context and output contract.

```python
review = task("aapl_order_activity", attempt="demo-01")
display_request(review)
```

## 3. Run

The SDK shows real activity and any questions here. Nothing selects a recommended reply automatically. Stopping this cell ends the local wait; keep its operation ID to reconnect.

```python
result = await run(epoch, review)
assert isinstance(result, ResearchCompleted), 'Answer or resume the pending question first.'
```

## 4. Read the complete answer

The session already displayed it. This cell reads every table page for Python use and checks the selected source and requested formats without printing a second report.

```python
answer = await read(epoch, result)
assert answer.reference.extension is not None
assert answer.reference.extension.source_execution_id == review.source_execution_id
assert result.response.content.declarations == review.request.outputs
rows = answer.tables["orders_by_side"]
assert len({row["side"] for row in rows}) == len(rows)
assert all(int(row["orders"]) >= 0 for row in rows)
assert len(answer.text["desk_note"].split()) <= 110
answer
```

## 5. Open the requested dashboard

Only the side-mix chart belongs on this dashboard. The table and note stay in the answer. The prepared environment supplies EPOCH_PORTAL_URL; this link selects the exact saved operation.

```python
report = await native_operation(epoch, result)
dashboard_url = dashboard(report)
```

## 6. Reconnect

A fresh client reads the same operation and saved answer. This does not submit a new research request.

```python
await epoch.aclose()
epoch = connect()
reconnected = await epoch.research_status(operation_id=review.operation_id)
assert isinstance(reconnected, ResearchCompleted)
assert reconnected.response == result.response
await epoch.aclose()
```
